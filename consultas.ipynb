**Consulta #1** 

Obtener el listado de proyectos asociados a cada disciplina, mostrando el nombre de la disciplina, el proyecto correspondiente y la cantidad de proyectos relacionados de mayor a menor


```python
select distinct
d.nombre_disciplina, p.nombre_proyecto,
count(p.id_proyecto) as cantidad_proyectos
from oferta as o 
inner join disciplina as d 
  on o.id_disciplina = d.id_disciplina
inner join estrategia as e
  on o.id_estrategia = e.id_estrategia
inner join proyecto as p 
  on e.id_proyecto = p.id_proyecto
group by
  d.id_disciplina,
  d.nombre_disciplina,
  p.id_proyecto,
  p.nombre_proyecto
order by 
  count(p.id_proyecto) DESC
```

**Consulta #2**

Contabilizar las ofertas registradas para cada combinación de disciplina y barrio de mayor a menor


```python
select
    d.nombre_disciplina,
    c.nombre_comuna,
    count(o.id_oferta) as cantidad_ofertas
from oferta as o
inner join disciplina as d
    on o.id_disciplina = d.id_disciplina
inner join barrio as b
    on o.id_barrio = b.id_barrio
inner join comuna as c
    on b.id_comuna = c.id_comuna
group by
    d.nombre_disciplina,
    c.nombre_comuna
order by
    count(o.id_oferta) DESC
```

**Consulta #3**

Promedio de personal formador y gestor por oferta en diferentes rangos de horarios



```python
select
    case
        when h.hora_inicio::time < '12:00' then 'mañana (<12:00)'
        when h.hora_inicio::time < '18:00' then 'tarde (12:00-18:00)'
        else 'noche (18:00-23:59)'
    end as horario,
    round(avg(
        (o.id_formador is not null)::int +
        (o.id_gestor is not null)::int
    ), 2) as promedio_personal
from oferta as o
join horario_oferta h using (id_oferta)
group by horario
order by min(h.hora_inicio);
```
